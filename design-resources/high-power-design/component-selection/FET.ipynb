# FET Transient Thermal Impedance Analysis

```python
from src.fet import FET, FETOperatingCondition

# Set of operating conditions for which peak current should calculated
cond1 = FETOperatingCondition("condition 1", pulse_width=20e-9, duty_cycle=-1, T_Jmax=110)
cond2 = FETOperatingCondition("condition 2", pulse_width=20e-9, duty_cycle=0.1, T_Jmax=110)
cond3 = FETOperatingCondition("condition 3", pulse_width=100e-9, duty_cycle=0.01, T_Jmax=110)
cond4 = FETOperatingCondition("condition 4", pulse_width=100e-9, duty_cycle=0.1, T_Jmax=110)

print("IMLT65R020M2H")
IMLT65R020M2H = FET(0.033, 0.33)
IMLT65R020M2H.Z_thJC[cond1] = 5e-3  # FETs have their transient thermal impedance assigned for each operating condition
IMLT65R020M2H.Z_thJC[cond2] = 3.8e-2
IMLT65R020M2H.Z_thJC[cond3] = 7e-3
IMLT65R020M2H.Z_thJC[cond4] = 3.8e-2
peak_currents = IMLT65R020M2H.get_peak_currents()  # works as expected
for cond, peak in peak_currents.items():
    print(f"Condition \"{cond.name}\":\t t_p={cond.pulse_width*1e9:.2f}ns\t D={cond.duty_cycle if cond.duty_cycle > 0 else "single pulse"}")
    print(f"\tPeak current: {peak:.6f} Amps")
print("\n")

print("GAN080650EBE")
GAN080650EBE = FET(0.135, 0.52)
GAN080650EBE.Z_thJC[cond1] = 2e-4
GAN080650EBE.Z_thJC[cond2] = 2.5e-4
GAN080650EBE.Z_thJC[cond3] = 5e-3
GAN080650EBE.Z_thJC[cond4] = 0.05
peak_currents = GAN080650EBE.get_peak_currents()
for cond, peak in peak_currents.items():
    print(f"Condition \"{cond.name}\":\t t_p={cond.pulse_width*1e9:.2f}ns\t D={cond.duty_cycle if cond.duty_cycle > 0 else "single pulse"}")
    print(f"\tPeak current: {peak:.6f} Amps")
print("\n")

print("IGLR65R200D2")
IGLR65R200D2 = FET(0.435, 3.7)
IGLR65R200D2.Z_thJC[cond1] = 1.7e-1
IGLR65R200D2.Z_thJC[cond2] = 5e-1
IGLR65R200D2.Z_thJC[cond3] = 2e-1
IGLR65R200D2.Z_thJC[cond4] = 5e-1
peak_currents = IGLR65R200D2.get_peak_currents()
for cond, peak in peak_currents.items():
    print(f"Condition \"{cond.name}\":\t t_p={cond.pulse_width*1e9:.2f}ns\t D={cond.duty_cycle if cond.duty_cycle > 0 else "single pulse"}")
    print(f"\tPeak current: {peak:.6f} Amps")
print("\n")

print("IMBG65R040M2H")
IMBG65R040M2H = FET(0.065, 0.76)
IMBG65R040M2H.Z_thJC[cond1] = 3e-2
IMBG65R040M2H.Z_thJC[cond2] = 1e-1
IMBG65R040M2H.Z_thJC[cond3] = 3.5e-2
IMBG65R040M2H.Z_thJC[cond4] = 1e-1
peak_currents = IMBG65R040M2H.get_peak_currents()
for cond, peak in peak_currents.items():
    print(f"Condition \"{cond.name}\":\t t_p={cond.pulse_width*1e9:.2f}ns\t D={cond.duty_cycle if cond.duty_cycle > 0 else "single pulse"}")
    print(f"\tPeak current: {peak:.6f} Amps")
print("\n")
```
